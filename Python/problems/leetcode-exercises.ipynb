#### Proposed solution for Two Sum problem
[LeetCode - Two Sum](https://leetcode.com/problems/two-sum/editorial/)

```polyglot-notebook
public class Solution 
{
    public int[] TwoSum(int[] nums, int target) 
    {
        if(nums.Length < 2 ||  nums.Length >  Math.Pow(10, 4))
            throw new ArgumentOutOfRangeException("nums", "Array length must be between 2 and 10^4");

        if(target < -Math.Pow(10, 9) ||  target >  Math.Pow(10, 9))
            throw new ArgumentOutOfRangeException("target", "Target must be between -10^4 and 10^9");

        for(int x = 0; x < nums.Length; x++)
        {
            int xnum = nums[x];

            for(int y = x + 1; y < nums.Length; y++) 
            {
                int ynum = nums[y];

                if(xnum + ynum == target)
                    return new int[] {x, y};
            }
        }
        return new int[] {-1, -1};
    }
}
var solution = new Solution();
var result = solution.TwoSum(new int[] {2,7,11,15,3,5,78,5,3,1,4,22,900,11,24,6,9,8,21,24,26,31,34,36,40,42,44,46,50,1}, 47);
Console.WriteLine($"[{string.Join(", ", result)}]");
```

#### Best solution for Two Sum problem (Claude AI)
[LeetCode - Two Sum](https://leetcode.com/problems/two-sum/editorial/)

```polyglot-notebook
public class Solution 
{
    public int[] TwoSum(int[] nums, int target) 
    {
        // Dictionary to store: value -> index
        var map = new Dictionary<int, int>();

        for (int i = 0; i < nums.Length; i++)
        {
            int complement = target - nums[i];
            
            Console.WriteLine($"Index: {i}, Number: {nums[i]}, Complement: {complement}, Map: [{string.Join(", ", map.Select(kv => kv.Key + "->" + kv.Value))}]");

            // Check if complement exists in map
            if (map.TryGetValue(complement, out int complementIndex))
            {
                return new int[] { complementIndex, i };
            }
            
            // Store current number and its index
            // Use indexer instead of Add to handle duplicates gracefully
            map[nums[i]] = i;
        }
        
        // Should never reach here per problem constraints
        throw new InvalidOperationException("No solution found");
    }
}
var solution = new Solution();
var result = solution.TwoSum(new int[] {2,7,11,15,3,5,78,5,3,1,4,22,900,11,24,6,9,8,21,24,26,31,34,36,40,42,44,46,50,1}, 51);
Console.WriteLine($"[{string.Join(", ", result)}]");
```
